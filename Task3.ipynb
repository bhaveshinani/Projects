```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# define a stock at $100
S = 100
# setting the parameters
K = 105  # strike price
premium = 3
S_T = np.arange(70,131,1)
```

```python
PnL_long = S_T - 100
PnL_Short = premium - np.maximum((S_T-K),0)
Net_PnL = PnL_long + P_L_Short
```

```python
x = S_T

plt.figure(figsize=(10,6))

plt.plot(x, PnL_long, linewidth=2,label= "Long stock")
plt.plot(x, PnL_Short, linewidth=2,label= "Short Call")
plt.plot(x, Net_PnL, linewidth=3, label=" Covered Call")

plt.axhline(0, color='black', linewidth=0.8)

Net_PnL = np.array(Net_PnL)

zero_idx = np.where(Net_PnL == 0)[0]
plt.scatter(x[zero_idx], Net_PnL[zero_idx], color='red', s=70, label='Break-even')

cap_level = np.max(Net_PnL)
plt.axvline(x=K, color='black', linestyle='--')
plt.axhline(cap_level, color='green', linestyle='--', label="capped upside")
plt.axvspan(K, S_T[-1], color='green', alpha=0.1, label="Capped Profit Zone")

plt.title("Long vs Short vs Covered")
plt.xlabel("Stock price")
plt.ylabel("Profit/Loss")
plt.legend()

plt.show()
```
