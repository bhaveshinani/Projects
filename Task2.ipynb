```python
import yfinance as yf
import numpy as np
import pandas as pd
import scipy.stats
from pandas.core.arrays import period
from datetime import datetime
```

```python
ticker = yf.Ticker("AAPL")
```

```python
expiry_date = ticker.options[0]
T = (datetime.strptime(expiry_date, "%Y-%m-%d").date() - datetime.today().date()).days
print(T)
```

```python
if T == 0:
  expiry_date = ticker.options[1]
  T = (datetime.strptime(expiry_date, "%Y-%m-%d").date() - datetime.today().date()).days
print(T)
T = T/365
T
```

```python
print(ticker.info)
```

```python
calls = ticker.option_chain(expiry_date).calls
calls
```

```python
puts = ticker.option_chain(expiry_date).puts
puts
```

```python
S = ticker.fast_info['lastPrice']
print(S)
```

```python
ticker_tnx = yf.Ticker("^TNX")
```

```python
data = ticker_tnx.history(period="1d")
r = data['Close'].iloc[0]/100
print(r)
```

```python
log_returns = np.log(ticker.history(period='18mo')['Close']/ticker.history(period='18mo')['Close'].shift(1))
log_returns = log_returns.tail(252)
sigma = log_returns.std() * np.sqrt(252)
sigma
```

```python
calls['ATM'] = abs(calls['strike'] - S)
K = calls[calls['ATM']==calls['ATM'].min()]['strike'].iloc[0]
calls[calls['ATM']==calls['ATM'].min()]
```

```python
puts['ATM'] = abs(puts['strike'] - S)
puts[puts['ATM']==puts['ATM'].min()]
```

```python
d_1 = (np.log(S/K)+(r+sigma**2/2)*T)/(sigma*T)
d_2 = d_1 - sigma*T
print(d_1,d_2)
```

```python
call_delta = scipy.stats.norm.pdf(d_1)
put_delta = call_delta-1
Gamma = scipy.stats.norm.cdf(d_1)/(S*sigma*np.sqrt(T))
call_theta = -(S*scipy.stats.norm.cdf(d_1)*sigma)/(2*np.sqrt(T)) - (r*K*((np.e)**(-r*T))*scipy.stats.norm.pdf(d_2))
put_theta = -(S*scipy.stats.norm.cdf(d_1)*sigma)/(2*np.sqrt(T)) + (r*K*((np.e)**(-r*T))*scipy.stats.norm.pdf(-d_2))
vega = S*np.sqrt(T)*scipy.stats.norm.cdf(d_1)
```

```python
df = pd.DataFrame({"Option Type": ['CAll', 'PUT'], "Delta":[call_delta,put_delta], "Gamma": [Gamma, Gamma], "Theta": [call_theta,put_theta], "Vega": [vega,vega]}).set_index("Option Type")
df
```
